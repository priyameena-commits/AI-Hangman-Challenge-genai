```python

import tkinter as tk
from tkinter import ttk, messagebox
import random

# ---------------- WORD BANK ----------------

WORD_BANK = {
    "Programming": {
        "PYTHON": [
            "This language is popular for artificial intelligence.",
            "It is named after a comedy group.",
            "It starts with P."
        ],
        "JAVASCRIPT": [
            "This language is widely used in web development.",
            "It runs in web browsers.",
            "Its name contains SCRIPT."
        ],
        "ALGORITHM": [
            "It is a step-by-step problem-solving method.",
            "It is important in data structures.",
            "It starts with ALGO."
        ],
        "DATABASE": [
            "It stores and organizes information.",
            "SQL is commonly used to manage it.",
            "It starts with DATA."
        ],
        "VARIABLE": [
            "It stores a value in a program.",
            "Its value can change.",
            "It starts with VAR."
        ]
    },
    "Science": {
        "GRAVITY": [
            "It keeps us on the ground.",
            "Planets exert this force.",
            "It starts with G."
        ],
        "OXYGEN": [
            "Humans need this gas to breathe.",
            "Its chemical symbol is O.",
            "It starts with O."
        ],
        "PLANET": [
            "Earth belongs to this category.",
            "It orbits a star.",
            "It starts with P."
        ],
        "ENERGY": [
            "It is needed to do work.",
            "It exists in different forms.",
            "It starts with E."
        ],
        "ATOM": [
            "It is a basic unit of matter.",
            "It contains a nucleus and electrons.",
            "It starts with A."
        ]
    },
    "Animals": {
        "ELEPHANT": [
            "It is a very large land animal.",
            "It has a long trunk.",
            "It starts with E."
        ],
        "GIRAFFE": [
            "It is famous for its long neck.",
            "It eats leaves from tall trees.",
            "It starts with G."
        ],
        "DOLPHIN": [
            "It is an intelligent aquatic mammal.",
            "It uses echolocation.",
            "It starts with D."
        ],
        "PENGUIN": [
            "It is a bird that cannot fly.",
            "It is an excellent swimmer.",
            "It starts with P."
        ],
        "CHEETAH": [
            "It is known for its incredible speed.",
            "It is a wild cat.",
            "It starts with C."
        ]
    },
    "Sports": {
        "CRICKET": [
            "It is very popular in India.",
            "Players use a bat and ball.",
            "It starts with C."
        ],
        "FOOTBALL": [
            "It is played by two teams.",
            "Players try to score goals.",
            "It starts with F."
        ],
        "TENNIS": [
            "Players use rackets.",
            "It is played across a net.",
            "It starts with T."
        ],
        "HOCKEY": [
            "Players use sticks.",
            "It can be played on ice or a field.",
            "It starts with H."
        ],
        "BADMINTON": [
            "Players use rackets and a shuttlecock.",
            "It is played across a net.",
            "It starts with B."
        ]
    }
}

# ---------------- COLORS ----------------

BG = "#101827"
PANEL = "#1B263B"
CARD = "#24334B"
TEXT = "#F1F5F9"
MUTED = "#A7B4C8"
TEAL = "#2DD4BF"
PURPLE = "#A78BFA"
RED = "#FB7185"
GREEN = "#4ADE80"
WHITE = "#FFFFFF"

MAX_ATTEMPTS = 6


# ---------------- GAME APPLICATION ----------------

class HangmanGame:

    def __init__(self, root):
        self.root = root
        self.root.title("AI Hangman Challenge")
        self.root.geometry("1000x760")
        self.root.minsize(800, 650)
        self.root.configure(bg=BG)

        self.score = 0
        self.round_number = 0
        self.rounds_won = 0
        self.game_over = False
        self.hint_index = 0

        self.category = tk.StringVar(value="Programming")

        self.build_interface()
        self.new_game()

    def build_interface(self):

        # Header
        header = tk.Frame(self.root, bg=BG)
        header.pack(fill="x", padx=24, pady=(18, 10))

        tk.Label(
            header, text="AI HANGMAN CHALLENGE",
            font=("Arial", 23, "bold"),
            bg=BG, fg=TEAL
        ).pack(anchor="w")

        tk.Label(
            header, text="Guess the hidden word. Unlock the clues.",
            font=("Arial", 11),
            bg=BG, fg=MUTED
        ).pack(anchor="w", pady=(4, 0))

        # Score information
        stats = tk.Frame(self.root, bg=BG)
        stats.pack(fill="x", padx=24, pady=(0, 14))

        self.score_label = tk.Label(
            stats, text="SCORE: 0",
            font=("Arial", 11, "bold"),
            bg=CARD, fg=TEAL, padx=15, pady=9
        )
        self.score_label.pack(side="left", padx=(0, 10))

        self.round_label = tk.Label(
            stats, text="ROUND: 1",
            font=("Arial", 11, "bold"),
            bg=CARD, fg=PURPLE, padx=15, pady=9
        )
        self.round_label.pack(side="left")

        # Main two-column area
        main = tk.Frame(self.root, bg=BG)
        main.pack(fill="both", expand=True, padx=24)

        # Left panel: Hangman drawing
        left = tk.Frame(main, bg=PANEL)
        left.pack(side="left", fill="both", expand=True, padx=(0, 12))

        tk.Label(
            left, text="HANGMAN",
            font=("Arial", 13, "bold"),
            bg=PANEL, fg=TEXT
        ).pack(pady=(16, 0))

        self.canvas = tk.Canvas(
            left, width=330, height=330,
            bg=PANEL, highlightthickness=0
        )
        self.canvas.pack(fill="both", expand=True, padx=8, pady=8)

        self.attempt_label = tk.Label(
            left, text="ATTEMPTS LEFT: 6 / 6",
            font=("Arial", 12, "bold"),
            bg=PANEL, fg=GREEN
        )
        self.attempt_label.pack(pady=(0, 8))

        self.status_label = tk.Label(
            left, text="Choose a letter to begin!",
            font=("Arial", 10),
            bg=PANEL, fg=MUTED,
            wraplength=300
        )
        self.status_label.pack(pady=(0, 16))

        # Right panel: Game controls
        right = tk.Frame(main, bg=PANEL)
        right.pack(side="right", fill="both", expand=True)

        tk.Label(
            right, text="GAME AREA",
            font=("Arial", 13, "bold"),
            bg=PANEL, fg=TEXT
        ).pack(anchor="w", padx=18, pady=(16, 10))

        category_row = tk.Frame(right, bg=PANEL)
        category_row.pack(fill="x", padx=18)

        tk.Label(
            category_row, text="Category",
            font=("Arial", 10),
            bg=PANEL, fg=MUTED
        ).pack(side="left")

        self.category_box = ttk.Combobox(
            category_row,
            textvariable=self.category,
            values=list(WORD_BANK.keys()),
            state="readonly",
            width=15
        )
        self.category_box.pack(side="right")
        self.category_box.bind(
            "<<ComboboxSelected>>",
            lambda event: self.new_game()
        )

        # Hidden word display
        self.word_label = tk.Label(
            right, text="",
            font=("Consolas", 25, "bold"),
            bg=PANEL, fg=TEAL
        )
        self.word_label.pack(pady=(25, 18))

        tk.Label(
            right, text="SELECT A LETTER",
            font=("Arial", 10, "bold"),
            bg=PANEL, fg=MUTED
        ).pack(anchor="w", padx=18, pady=(0, 8))

        # Alphabet buttons
        self.letter_frame = tk.Frame(right, bg=PANEL)
        self.letter_frame.pack(fill="x", padx=12)

        self.letter_buttons = {}

        for index, letter in enumerate("ABCDEFGHIJKLMNOPQRSTUVWXYZ"):
            button = tk.Button(
                self.letter_frame,
                text=letter,
                font=("Arial", 10, "bold"),
                width=3,
                bg=CARD,
                fg=TEXT,
                activebackground=TEAL,
                activeforeground=BG,
                relief="flat",
                cursor="hand2",
                command=lambda ch=letter: self.guess_letter(ch)
            )
            button.grid(
                row=index // 7,
                column=index % 7,
                padx=3, pady=3, sticky="nsew"
            )
            self.letter_buttons[letter] = button

        for col in range(7):
            self.letter_frame.grid_columnconfigure(
                col, weight=1
            )

        # Hint card
        hint_card = tk.Frame(right, bg=CARD)
        hint_card.pack(fill="x", padx=18, pady=(18, 10))

        tk.Label(
            hint_card, text="PROGRESSIVE HINT",
            font=("Arial", 10, "bold"),
            bg=CARD, fg=PURPLE
        ).pack(anchor="w", padx=12, pady=(10, 5))

        self.hint_label = tk.Label(
            hint_card,
            text="Need help? Reveal a hint!",
            font=("Arial", 10),
            bg=CARD, fg=TEXT,
            wraplength=340,
            justify="left"
        )
        self.hint_label.pack(
            anchor="w", padx=12, pady=(0, 10)
        )

        self.hint_button = tk.Button(
            right, text="GET NEXT HINT  (-5 POINTS)",
            font=("Arial", 10, "bold"),
            bg=PURPLE, fg=BG,
            activebackground="#C4B5FD",
            relief="flat", cursor="hand2",
            command=self.get_hint
        )
        self.hint_button.pack(
            fill="x", padx=18, pady=(0, 12)
        )

        # Bottom buttons
        footer = tk.Frame(self.root, bg=BG)
        footer.pack(fill="x", padx=24, pady=18)

        tk.Button(
            footer, text="NEW GAME",
            font=("Arial", 11, "bold"),
            bg=TEAL, fg=BG,
            activebackground="#99F6E4",
            relief="flat", cursor="hand2",
            padx=18, pady=10,
            command=self.new_game
        ).pack(side="left")

        tk.Button(
            footer, text="EXIT",
            font=("Arial", 11, "bold"),
            bg=CARD, fg=TEXT,
            activebackground=RED,
            relief="flat", cursor="hand2",
            padx=18, pady=10,
            command=self.root.destroy
        ).pack(side="right")

    # ---------------- HANGMAN DRAWING ----------------

    def draw_hangman(self):
        self.canvas.delete("all")

        # Gallows
        self.canvas.create_line(
            45, 290, 160, 290,
            fill=MUTED, width=5
        )
        self.canvas.create_line(
            85, 290, 85, 35,
            fill=MUTED, width=5
        )
        self.canvas.create_line(
            85, 35, 205, 35,
            fill=MUTED, width=5
        )
        self.canvas.create_line(
            205, 35, 205, 65,
            fill=MUTED, width=4
        )

        # Draw one body part per wrong guess
        if self.wrong_count >= 1:
            self.canvas.create_oval(
                180, 65, 230, 115,
                outline=RED, width=4
            )

        if self.wrong_count >= 2:
            self.canvas.create_line(
                205, 115, 205, 195,
                fill=RED, width=4
            )

        if self.wrong_count >= 3:
            self.canvas.create_line(
                205, 135, 170, 165,
                fill=RED, width=4
            )

        if self.wrong_count >= 4:
            self.canvas.create_line(
                205, 135, 240, 165,
                fill=RED, width=4
            )

        if self.wrong_count >= 5:
            self.canvas.create_line(
                205, 195, 175, 240,
                fill=RED, width=4
            )

        if self.wrong_count >= 6:
            self.canvas.create_line(
                205, 195, 235, 240,
                fill=RED, width=4
            )

    # ---------------- GAME LOGIC ----------------

    def new_game(self):
        self.round_number += 1
        self.game_over = False
        self.hint_index = 0
        self.wrong_count = 0
        self.guessed_letters = set()
        self.wrong_letters = set()

        category_words = WORD_BANK[self.category.get()]
        self.word = random.choice(list(category_words.keys()))
        self.hints = category_words[self.word]

        self.word_label.config(text=self.get_display_word())
        self.hint_label.config(
            text="Need help? Click GET NEXT HINT!"
        )
        self.status_label.config(
            text="Choose a letter to begin!",
            fg=MUTED
        )

        self.attempt_label.config(
            text=f"ATTEMPTS LEFT: {MAX_ATTEMPTS} / {MAX_ATTEMPTS}",
            fg=GREEN
        )

        self.hint_button.config(state="normal")

        for button in self.letter_buttons.values():
            button.config(
                state="normal",
                bg=CARD,
                fg=TEXT
            )

        self.draw_hangman()
        self.update_stats()

    def get_display_word(self):
        return " ".join(
            letter if letter in self.guessed_letters else "_"
            for letter in self.word
        )

    def update_stats(self):
        self.score_label.config(text=f"SCORE: {self.score}")
        self.round_label.config(text=f"ROUND: {self.round_number}")

    def guess_letter(self, letter):
        if self.game_over:
            return

        self.letter_buttons[letter].config(state="disabled")

        if letter in self.word:
            self.guessed_letters.add(letter)

            self.letter_buttons[letter].config(
                bg=TEAL, fg=BG, disabledforeground=BG
            )

            self.score += 10
            self.status_label.config(
                text=f"Great! {letter} is in the word.",
                fg=GREEN
            )

        else:
            self.wrong_letters.add(letter)
            self.wrong_count += 1

            self.letter_buttons[letter].config(
                bg=RED, fg=BG, disabledforeground=BG
            )

            self.status_label.config(
                text=f"Oops! {letter} is not in the word.",
                fg=RED
            )

        self.word_label.config(text=self.get_display_word())
        self.draw_hangman()

        remaining = MAX_ATTEMPTS - self.wrong_count

        self.attempt_label.config(
            text=f"ATTEMPTS LEFT: {remaining} / {MAX_ATTEMPTS}",
            fg=GREEN if remaining > 2 else RED
        )

        self.update_stats()

        if all(letter in self.guessed_letters for letter in self.word):
            self.score += 20
            self.rounds_won += 1
            self.end_game(True)

        elif self.wrong_count >= MAX_ATTEMPTS:
            self.end_game(False)

        self.update_stats()

    def get_hint(self):
        if self.game_over:
            return

        if self.hint_index < len(self.hints):
            hint = self.hints[self.hint_index]
            self.hint_index += 1

            self.score = max(0, self.score - 5)

            self.hint_label.config(
                text=f"Hint {self.hint_index}: {hint}"
            )

            self.update_stats()

            if self.hint_index == len(self.hints):
                self.hint_button.config(
                    text="NO MORE HINTS",
                    state="disabled"
                )
        else:
            self.hint_label.config(text="No more hints available!")

    def end_game(self, won):
        self.game_over = True

        for button in self.letter_buttons.values():
            button.config(state="disabled")

        self.hint_button.config(state="disabled")

        if won:
            self.word_label.config(text=" ".join(self.word))
            self.status_label.config(
                text="Excellent! You solved the word. +20 bonus!",
                fg=GREEN
            )
            messagebox.showinfo(
                "You Won!",
                f"Congratulations!\n\n"
                f"The word was: {self.word}\n"
                f"Round bonus: 20 points\n"
                f"Total score: {self.score}"
            )
        else:
            self.word_label.config(text=" ".join(self.word))
            self.status_label.config(
                text="Game over! The correct word is revealed.",
                fg=RED
            )
            messagebox.showinfo(
                "Game Over",
                f"You ran out of attempts.\n\n"
                f"The word was: {self.word}\n"
                f"Total score: {self.score}"
            )


# ---------------- START APPLICATION ----------------

if __name__ == "__main__":
    root = tk.Tk()
    app = HangmanGame(root)
    root.mainloop()
```
